```python
import os
from ultralytics import YOLO
import torch
import shutil
from IPython.display import FileLink

# Check if GPU is available and set the device accordingly
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Initialize the YOLOv8 model (you can change this to yolov8s, yolov8m, etc., depending on your resources)
model = YOLO('yolov8n.pt')  # Use the appropriate YOLO version, such as yolov8n.pt or yolov8s.pt

# Path to the Blood_Violence-2 dataset's YAML file
data_yaml_path = '/kaggle/working/Blood_Violence-2/data.yaml'  # Path to the Blood_Violence-2 dataset's data.yaml file


# Output directory to save the trained model and results
output_dir = '/kaggle/working/runs/train'  # Directory to store training results

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Train the model
model.train(
    data=data_yaml_path,           # Path to the data.yaml file for your dataset
    epochs=25,                     # Number of epochs to train (adjust as needed)
    imgsz=640,                     # Image size (640x640 is standard)
    batch=16,                      # Batch size (adjust depending on GPU memory)
    name='blood_violence_model',    # Name for the experiment
    save=True,                     # Save the model weights during training
    project=output_dir,            # Directory to store training results
    exist_ok=True,                 # Allow overwriting existing experiments
    device=device                  # Use GPU if available
)

# After training, best.pt and last.pt will be saved automatically.

# Reload the best or last model after training (you can choose either one)
best_model_path = os.path.join(output_dir, 'blood_violence_model', 'weights', 'best.pt')
last_model_path = os.path.join(output_dir, 'blood_violence_model', 'weights', 'last.pt')

# You can reload and save the best or last model
model = YOLO(best_model_path)  # Load the best model after training

# Save the final model in a specific directory (if needed)
final_model_path = '/kaggle/working/final_blood_violence_model.pt'
model.save(final_model_path)

# Move the final model to an accessible location for downloading after training
final_model_path_local = '/kaggle/working/final_blood_violence_model.pt'
shutil.move(final_model_path, final_model_path_local)

# Optionally, create a downloadable link for the model file
download_link = FileLink(final_model_path_local)

# Display the download link
print(f"Download your trained model from the following link: {download_link}")
```

```python
import os
from ultralytics import YOLO
import torch
import shutil
from IPython.display import FileLink

# Check if GPU is available and set the device accordingly
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Path to the pre-trained weights
pretrained_weights_path = '/kaggle/input/yolo/other/default/1/weights/best.pt'  # Path to your pre-trained model weights

# Initialize the YOLO model with the pre-trained weights (best.pt)
model = YOLO(pretrained_weights_path)  # Load pre-trained weights for further training

# Path to the Bottels2-1 dataset's YAML file
data_yaml_path = '/kaggle/working/Bottels2-1/data.yaml'  # Path to the data.yaml file for your dataset

# Output directory to save the trained model and results
output_dir = '/kaggle/working/runs/train'  # Directory to store training results

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Train the model further with your custom dataset
model.train(
    data=data_yaml_path,           # Path to the data.yaml file for your dataset
    epochs=25,                     # Number of epochs to train (adjust as needed)
    imgsz=640,                     # Image size (640x640 is standard)
    batch=16,                      # Batch size (adjust depending on GPU memory)
    name='Bottels2-1_finetuned',    # Name for the experiment
    save=True,                     # Save the model weights during training
    project=output_dir,            # Directory to store training results
    exist_ok=True,                 # Allow overwriting existing experiments
    device=device                  # Use GPU if available
)

# After training, best.pt and last.pt will be saved automatically.

# Reload the best or last model after training (you can choose either one)
best_model_path = os.path.join(output_dir, 'Bottels2-1_finetuned', 'weights', 'best.pt')
last_model_path = os.path.join(output_dir, 'Bottels2-1_finetuned', 'weights', 'last.pt')

# You can reload and save the best or last model
model = YOLO(best_model_path)  # Load the best model after training

# Save the final model in a specific directory (if needed)
final_model_path = '/kaggle/working/final_model_Bottels2-1_finetuned.pt'
model.save(final_model_path)

# Move the final model to an accessible location for downloading after training
final_model_path_local = '/kaggle/working/final_model_Bottels2-1_finetuned.pt'
shutil.move(final_model_path, final_model_path_local)

# Optionally, create a downloadable link for the model file
download_link = FileLink(final_model_path_local)

# Display the download link
print(f"Download your trained model from the following link: {download_link}")
```

```python
import os
import requests
from ultralytics import YOLO
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt

# Path to the trained model
model_path = '/kaggle/working/runs/train/Bottels2-1_finetuned/weights/best.pt'

# Load the trained model
model = YOLO(model_path)

# URL of the image (replace this with the image URL you want to use)
image_url = 'https://www.livemint.com/lm-img/img/2024/12/27/1600x900/GfuoPw7WcAA9-6b_1735267294793_1735267323768.jpg'  # Replace with your image URL

# Step 1: Download the image from the URL
response = requests.get(image_url)
if response.status_code == 200:
    img_data = response.content
    print("Image downloaded successfully.")
else:
    print("Failed to download the image.")
    img_data = None

# Step 2: Load the image into PIL and then convert it to a format suitable for YOLO
if img_data:
    img = Image.open(BytesIO(img_data))  # Convert the downloaded bytes into a PIL image
    
    # Resize the image to 640x640 (standard for YOLO)
    img = img.resize((640, 640))
    
    # Optional: Save the image locally if you want
    image_path = '/kaggle/working/downloaded_image_resized.jpg'
    img.save(image_path)
    print(f"Resized image saved as {image_path}")
    
    # Step 3: Perform inference with the YOLO model
    results = model.predict(source=image_path, conf=0.3)  # Lower confidence threshold to 0.3
    
    # Check if predictions are made
    if len(results) > 0 and len(results[0].boxes) > 0:
        print(f"Found {len(results[0].boxes)} detections.")
    else:
        print("No detections found.")
    
    # Step 4: Plot the result with bounding boxes
    result_img = results[0].plot()  # Get the image with bounding boxes
    plt.figure(figsize=(10, 10))
    plt.imshow(result_img)
    plt.axis('off')
    plt.show()

    # Optionally, save the result to a file
    result_image_path = '/kaggle/working/predicted_image_resized.jpg'
    result_img = results[0].plot()  # Re-run plotting to save the predicted image
    plt.imsave(result_image_path, result_img)
    print(f"Prediction saved to: {result_image_path}")
```

```python
import os
from ultralytics import YOLO
import torch
import shutil
from IPython.display import FileLink

# Check if GPU is available and set the device accordingly
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Path to the pre-trained weights
pretrained_weights_path = '/kaggle/input/yolo/other/default/1/weights/last.pt'  # Path to your pre-trained model weights

# Initialize the YOLO model with the pre-trained weights (best.pt)
model = YOLO(pretrained_weights_path)  # Load pre-trained weights for further training

# Path to the Bottels2-1 dataset's YAML file
data_yaml_path = '/kaggle/working/Bottels2-1/data.yaml'  # Path to the data.yaml file for your dataset

# Output directory to save the trained model and results
output_dir = '/kaggle/working/runs/train'  # Directory to store training results

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Train the model further with your custom dataset
model.train(
    data=data_yaml_path,           # Path to the data.yaml file for your dataset
    epochs=100,                     # Number of epochs to train (adjust as needed)
    imgsz=640,                     # Image size (640x640 is standard)
    batch=16,                      # Batch size (adjust depending on GPU memory)
    name='Bottels2-1_finetuned_2',    # Name for the experiment
    save=True,                     # Save the model weights during training
    project=output_dir,            # Directory to store training results
    exist_ok=True,                 # Allow overwriting existing experiments
    device=device                  # Use GPU if available
)

# After training, best.pt and last.pt will be saved automatically.

# Reload the best or last model after training (you can choose either one)
best_model_path = os.path.join(output_dir, 'Bottels2-1_finetuned', 'weights', 'best.pt')
last_model_path = os.path.join(output_dir, 'Bottels2-1_finetuned', 'weights', 'last.pt')

# You can reload and save the best or last model
model = YOLO(best_model_path)  # Load the best model after training

# Save the final model in a specific directory (if needed)
final_model_path = '/kaggle/working/final_model_Bottels2-1_finetuned_2.pt'
model.save(final_model_path)

# Move the final model to an accessible location for downloading after training
final_model_path_local = '/kaggle/working/final_model_Bottels2-1_finetuned_2.pt'
shutil.move(final_model_path, final_model_path_local)

# Optionally, create a downloadable link for the model file
download_link = FileLink(final_model_path_local)

# Display the download link
print(f"Download your trained model from the following link: {download_link}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
from matplotlib import pyplot as plt

# Path to your trained model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file (replace with your video path)
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Optionally, create an output video file to save the predictions
output_video_path = '/kaggle/working//output_video.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

# Loop through the video frame by frame
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    
    # Resize frame for YOLO model (640x640)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the frame
    results = model.predict(source=resized_frame, conf=0.3)  # Lower confidence threshold to 0.3

    # Check if predictions are made
    if len(results) > 0 and len(results[0].boxes) > 0:
        print(f"Found {len(results[0].boxes)} detections.")
    else:
        print("No detections found.")
    
    # Plot the result with bounding boxes on the frame
    result_frame = results[0].plot()  # This gives the frame with bounding boxes

    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Write the frame to the output video
    out.write(result_frame)

    # Display the frame with bounding boxes (optional, can be commented out)
    cv2.imshow("Frame", result_frame)

    # Press 'q' to quit the video preview
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

print(f"Prediction video saved to: {output_video_path}")
```

```python
# import os
from ultralytics import YOLO
import torch

# Check if GPU is available and set the device accordingly
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Initialize the YOLOv8 model (you can use yolov8n, yolov8s, yolov8m, etc., depending on your resources)
model = YOLO('yolov8n.pt')  # Use a smaller model for faster training or change to yolov8s.pt or yolov8m.pt for more accuracy

# Path to the License-Plate-Recognition-2 dataset's YAML file
data_yaml_path = '/kaggle/working/License-Plate-Recognition-2/data.yaml'  # Path to your dataset's data.yaml file

# Output directory to save the trained model and results
output_dir = '/kaggle/working/runs/train'  # Directory to store training results

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Train the model
model.train(
    data=data_yaml_path,           # Path to the data.yaml file for your dataset
    epochs=25,                     # Number of epochs to train (adjust as needed)
    imgsz=640,                     # Image size (640x640 is standard)
    batch=16,                      # Batch size (adjust depending on GPU memory)
    name='license_plate_model',    # Name for the experiment
    save=True,                     # Save the model weights during training
    project=output_dir,            # Directory to store training results
    exist_ok=True,                 # Allow overwriting existing experiments
    device=device                  # Use GPU if available
)

# After training, best.pt and last.pt will be saved automatically.

# Reload the best or last model after training (you can choose either one)
best_model_path = os.path.join(output_dir, 'license_plate_model', 'weights', 'best.pt')
last_model_path = os.path.join(output_dir, 'license_plate_model', 'weights', 'last.pt')

# You can reload and save the best or last model
model = YOLO(best_model_path)  # Load the best model after training

# Save the final model in a specific directory (if needed)
final_model_path = '/kaggle/working/final_license_plate_model.pt'
model.save(final_model_path)

# Move the final model to an accessible location for downloading after training
final_model_path_local = '/kaggle/working/final_license_plate_model.pt'
import shutil
shutil.move(final_model_path, final_model_path_local)

# Optionally, create a downloadable link for the model file
from IPython.display import FileLink
download_link = FileLink(final_model_path_local)

# Display the download link
print(f"Download your trained model from the following link: {download_link}")
```

```python
import torch
import cv2
import numpy as np
import os

# Ensure the model file exists before loading
model_path = '/kaggle/input/blood_model/other/default/1/weights/best.pt'

if not os.path.exists(model_path):
    raise FileNotFoundError(f"Model file not found at {model_path}")

# Clear any existing YOLOv5 cache (optional but can resolve issues)
torch.hub._validate_not_a_forked_repo = True  # Avoid issues with forked repositories

# Load the YOLOv5 model (this will download and load the correct model architecture and weights)
model = torch.hub.load('ultralytics/yolov5', 'custom', path=model_path, force_reload=True)

# Load the input image (can be a local path or URL)
image_path = "https://www.shutterstock.com/image-photo/blood-splashes-splatter-trash-background-260nw-207602554.jpg"
image = cv2.imread(image_path)

# Convert the image to RGB for YOLOv5 inference
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Perform inference (detection)
results = model(image_rgb)

# Get the predictions (bounding boxes, confidence scores, and class labels)
predictions = results.xywh[0].cpu().numpy()  # predictions[0] gives detections for the image
boxes = predictions[:, :4]  # x1, y1, x2, y2
confidences = predictions[:, 4]  # confidence scores
class_ids = predictions[:, 5].astype(int)  # detected class IDs

# Set the threshold for the confidence score to blur only the correctly detected areas
confidence_threshold = 0.5  # Adjust this threshold as needed

# Loop over the detections and blur the regions with high confidence
for box, confidence in zip(boxes, confidences):
    if confidence >= confidence_threshold:
        x1, y1, x2, y2 = box.astype(int)

        # Extract the detected region from the image
        detected_region = image[y1:y2, x1:x2]

        # Apply Gaussian blur to the detected region
        blurred_region = cv2.GaussianBlur(detected_region, (15, 15), 0)

        # Replace the original detected region with the blurred one
        image[y1:y2, x1:x2] = blurred_region

# Convert the image back to BGR for OpenCV
image_bgr = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

# Save or display the image
cv2.imwrite("blurred_output.jpg", image_bgr)  # Save the output image
cv2.imshow("Blurred Output", image_bgr)  # Optionally display the image
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import os
import requests
from ultralytics import YOLO
from PIL import Image, ImageFilter
import numpy as np
import cv2
import matplotlib.pyplot as plt
from io import BytesIO

# Path to the trained model (replace with actual .pt model file)
model_path = '/kaggle/input/blood_model/other/default/1/weights/best.pt'

# Load the trained model
model = YOLO(model_path)

# URL of the image (replace this with the image URL you want to use)
image_url = 'https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRk2F2cqnErjrfa2HZioN50NLmXqM_AZdnoKg&s'  # Replace with your image URL

# Step 1: Download the image from the URL
response = requests.get(image_url)
if response.status_code == 200:
    img_data = response.content
    print("Image downloaded successfully.")
else:
    print("Failed to download the image.")
    img_data = None

# Step 2: Load the image into PIL and then convert it to a format suitable for YOLO
if img_data:
    img = Image.open(BytesIO(img_data))  # Convert the downloaded bytes into a PIL image
    
    # Resize the image to 640x640 (standard for YOLO)
    img = img.resize((640, 640))
    
    # Optional: Save the image locally if you want
    image_path = '/kaggle/working/downloaded_image_resized.jpg'
    img.save(image_path)
    print(f"Resized image saved as {image_path}")
    
    # Step 3: Perform inference with the YOLO model
    results = model.predict(source=image_path, conf=0.3)  # Lower confidence threshold to 0.3
    
    # Check if predictions are made
    if len(results[0].boxes) > 0:
        print(f"Found {len(results[0].boxes)} detections.")
    else:
        print("No detections found.")
    
    # Step 4: Apply mosaic (pixelation) to the detected regions
    img_array = np.array(img)  # Convert image to numpy array for manipulation
    
    for box in results[0].boxes:
        # Extract bounding box coordinates
        x1, y1, x2, y2 = box.xyxy[0].int().tolist()  # Get the coordinates of the bounding box
        print(f"Detected bounding box: {x1}, {y1}, {x2}, {y2}")
        
        # Ensure the coordinates are valid and within image bounds
        x1, y1, x2, y2 = max(0, x1), max(0, y1), min(img_array.shape[1], x2), min(img_array.shape[0], y2)

        # Extract the detected region from the image
        detected_region = img_array[y1:y2, x1:x2]
        
        # Apply mosaic effect (pixelation)
        pixelated_region = cv2.resize(detected_region, (10, 10), interpolation=cv2.INTER_LINEAR)
        pixelated_region = cv2.resize(pixelated_region, (x2 - x1, y2 - y1), interpolation=cv2.INTER_NEAREST)
        
        # Replace the original detected region with the pixelated one
        img_array[y1:y2, x1:x2] = pixelated_region

    # Convert back the array to an image
    pixelated_img = Image.fromarray(img_array)

    # Step 5: Plot the result with bounding boxes
    result_img = results[0].plot()  # Get the image with bounding boxes
    plt.figure(figsize=(10, 10))
    plt.imshow(pixelated_img)  # Display the pixelated image with bounding boxes
    plt.axis('off')
    plt.show()

    # Optionally, save the result to a file
    result_image_path = '/kaggle/working/pixelated_prediction_image.jpg'
    plt.imsave(result_image_path, np.array(pixelated_img))  # Save the final pixelated image
    print(f"Pixelated prediction saved to: {result_image_path}")
```

```python
import os
import requests
from ultralytics import YOLO
from PIL import Image
import numpy as np
import cv2
import matplotlib.pyplot as plt
from io import BytesIO

# Path to the trained smoke detection model (replace with actual .pt model file)
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Load the trained model
model = YOLO(model_path)

# URL of the image (replace this with the image URL you want to use)
image_url = 'https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTpr-SaBTzysjdS3EJ2HoNQhCU2KnLvAHAtiw&s'  # Replace with your image URL

# Step 1: Download the image from the URL
response = requests.get(image_url)
if response.status_code == 200:
    img_data = response.content
    print("Image downloaded successfully.")
else:
    print("Failed to download the image.")
    img_data = None

# Step 2: Load the image into PIL and then convert it to a format suitable for YOLO
if img_data:
    img = Image.open(BytesIO(img_data))  # Convert the downloaded bytes into a PIL image
    
    # Resize the image to 640x640 (standard for YOLO)
    img = img.resize((640, 640))
    
    # Optional: Save the image locally if you want
    image_path = '/kaggle/working/downloaded_image_resized.jpg'
    img.save(image_path)
    print(f"Resized image saved as {image_path}")
    
    # Step 3: Perform inference with the YOLO model
    results = model.predict(source=image_path, conf=0.3)  # Lower confidence threshold to 0.3
    
    # Check if predictions are made
    if len(results[0].boxes) > 0:
        print(f"Found {len(results[0].boxes)} detections.")
    else:
        print("No detections found.")
    
    # Step 4: Apply mosaic (pixelation) to the detected regions
    img_array = np.array(img)  # Convert image to numpy array for manipulation
    
    for box in results[0].boxes:
        # Extract bounding box coordinates
        x1, y1, x2, y2 = box.xyxy[0].int().tolist()  # Get the coordinates of the bounding box
        print(f"Detected bounding box: {x1}, {y1}, {x2}, {y2}")
        
        # Ensure the coordinates are valid and within image bounds
        x1, y1, x2, y2 = max(0, x1), max(0, y1), min(img_array.shape[1], x2), min(img_array.shape[0], y2)

        # Extract the detected region from the image
        detected_region = img_array[y1:y2, x1:x2]
        
        # Apply mosaic effect (pixelation)
        pixelated_region = cv2.resize(detected_region, (10, 10), interpolation=cv2.INTER_LINEAR)
        pixelated_region = cv2.resize(pixelated_region, (x2 - x1, y2 - y1), interpolation=cv2.INTER_NEAREST)
        
        # Replace the original detected region with the pixelated one
        img_array[y1:y2, x1:x2] = pixelated_region

    # Convert back the array to an image
    pixelated_img = Image.fromarray(img_array)

    # Step 5: Plot the result with bounding boxes
    result_img = results[0].plot()  # Get the image with bounding boxes
    plt.figure(figsize=(10, 10))
    plt.imshow(pixelated_img)  # Display the pixelated image with bounding boxes
    plt.axis('off')
    plt.show()

    # Optionally, save the result to a file
    result_image_path = '/kaggle/working/pixs_prediction_image.jpg'
    plt.imsave(result_image_path, np.array(pixelated_img))  # Save the final pixelated image
    print(f"Pixelated prediction saved to: {result_image_path}")
```

```python
import os
import requests
from ultralytics import YOLO
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from io import BytesIO
import cv2  # Importing OpenCV to apply Gaussian blur

# Path to the trained smoke detection model (replace with actual .pt model file)
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Load the trained model
model = YOLO(model_path)

# URL of the image (replace this with the image URL you want to use)
image_url = 'https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTpr-SaBTzysjdS3EJ2HoNQhCU2KnLvAHAtiw&s'  # Replace with your image URL

# Step 1: Download the image from the URL
response = requests.get(image_url)
if response.status_code == 200:
    img_data = response.content
    print("Image downloaded successfully.")
else:
    print("Failed to download the image.")
    img_data = None

# Step 2: Load the image into PIL and then convert it to a format suitable for YOLO
if img_data:
    img = Image.open(BytesIO(img_data))  # Convert the downloaded bytes into a PIL image
    
    # Resize the image to 640x640 (standard for YOLO)
    img = img.resize((640, 640))
    
    # Optional: Save the image locally if you want
    image_path = '/kaggle/working/downloaded_image_resized.jpg'
    img.save(image_path)
    print(f"Resized image saved as {image_path}")
    
    # Step 3: Perform inference with the YOLO model
    results = model.predict(source=image_path, conf=0.3)  # Lower confidence threshold to 0.3
    
    # Check if predictions are made
    if len(results[0].boxes) > 0:
        print(f"Found {len(results[0].boxes)} detections.")
    else:
        print("No detections found.")
    
    # Step 4: Apply strong Gaussian blur to the detected regions
    img_array = np.array(img)  # Convert image to numpy array for manipulation
    
    for box in results[0].boxes:
        # Extract bounding box coordinates
        x1, y1, x2, y2 = box.xyxy[0].int().tolist()  # Get the coordinates of the bounding box
        print(f"Detected bounding box: {x1}, {y1}, {x2}, {y2}")
        
        # Ensure the coordinates are valid and within image bounds
        x1, y1, x2, y2 = max(0, x1), max(0, y1), min(img_array.shape[1], x2), min(img_array.shape[0], y2)

        # Extract the detected region from the image
        detected_region = img_array[y1:y2, x1:x2]
        
        # Apply a stronger Gaussian blur to the detected region (increased kernel size for stronger blur)
        blurred_region = cv2.GaussianBlur(detected_region, (35, 35), 0)  # Stronger blur
        
        # Replace the original detected region with the blurred one
        img_array[y1:y2, x1:x2] = blurred_region

    # Convert back the array to an image
    blurred_img = Image.fromarray(img_array)

    # Step 5: Plot the result with bounding boxes
    result_img = results[0].plot()  # Get the image with bounding boxes
    plt.figure(figsize=(10, 10))
    plt.imshow(blurred_img)  # Display the blurred image with bounding boxes
    plt.axis('off')
    plt.show()

    # Optionally, save the result to a file
    result_image_path = '/kaggle/working/SMOKEimage_blurred.jpg'
    plt.imsave(result_image_path, np.array(blurred_img))  # Save the final blurred image
    print(f"Blurred prediction saved to: {result_image_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
from matplotlib import pyplot as plt

# Path to your trained model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file (replace with your video path)
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Optionally, create an output video file to save the predictions
output_video_path = '/kaggle/working/output_video.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

# Loop through the video frame by frame
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    
    # Resize frame for YOLO model (640x640)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the frame
    results = model.predict(source=resized_frame, conf=0.3)  # Lower confidence threshold to 0.3

    # Check if predictions are made
    if len(results) > 0 and len(results[0].boxes) > 0:
        print(f"Found {len(results[0].boxes)} detections.")
    else:
        print("No detections found.")
    
    # Plot the result with bounding boxes on the frame
    result_frame = results[0].plot()  # This gives the frame with bounding boxes




    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Write the frame to the output video
    out.write(result_frame)

    # Display the frame with bounding boxes (optional, can be commented out)
    cv2.imshow("Frame", result_frame)

    # Press 'q' to quit the video preview
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

print(f"Prediction video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np

# Path to your trained model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file (replace with your video path)
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)
if fps == 0:
    print("Error: Unable to fetch FPS.")
else:
    print(f"Video FPS: {fps}")

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Output video path
output_video_path = '/kaggle/working/outputvideo.mp4'

# Set the codec to MJPG for better compatibility
fourcc = cv2.VideoWriter_fourcc(*'MJPG')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

# Loop through the video frame by frame
frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Resize frame for YOLO model (640x640)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the frame
    results = model.predict(source=resized_frame, conf=0.3)

    # Check if predictions are made
    if len(results) > 0 and len(results[0].boxes) > 0:
        print(f"Frame {frame_count}: Found {len(results[0].boxes)} detections.")
    else:
        print(f"Frame {frame_count}: No detections found.")

    # Plot the result with bounding boxes on the frame
    result_frame = results[0].plot()

    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Write the frame to the output video
    out.write(result_frame)

    # Optionally show the frame (for debugging purposes)
    cv2.imshow("Frame", result_frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

    frame_count += 1

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

print(f"Prediction video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time  # Import time module to get timestamps

# Path to your trained model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file (replace with your video path)
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Desired FPS (10 FPS in this case)
desired_fps = 10
frame_interval = int(fps // desired_fps)  # Calculate how many frames to skip

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

# Prepare output video (optional, if you want to save the output)
output_video_path = 'output_video.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, desired_fps, (frame_width, frame_height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Skip frames based on the desired FPS
    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    # Debug print to track frame count and interval
    print(f"Processing frame {frame_count}")

    # Resize the frame for YOLO model to 640x640 (for inference)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the resized frame
    results = model.predict(source=resized_frame, conf=0.3)  # Reduce confidence threshold

    # Check if predictions are made
    if len(results) > 0 and len(results[0].boxes) > 0:
        print(f"Frame {frame_count}: Found {len(results[0].boxes)} detections.")
    else:
        print(f"Frame {frame_count}: No detections found.")

    # Plot the result with bounding boxes on the original frame (not resized)
    result_frame = results[0].plot()  # This plots on the original frame (with bounding boxes)

    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Get the current timestamp in seconds based on frame count
    timestamp_seconds = frame_count / fps  # Time in seconds based on frame count and fps
    timestamp = time.strftime('%H:%M:%S', time.gmtime(timestamp_seconds))  # Convert seconds to HH:MM:SS format

    # Add the timestamp to the frame
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(result_frame, timestamp, (10, 30), font, 1, (255, 255, 255), 2, cv2.LINE_AA)

    # Add the disclaimer text at the center bottom
    disclaimer_text = "Smoking is injurious to health"
    text_size = cv2.getTextSize(disclaimer_text, font, 1, 2)[0]  # Get text size
    text_width = text_size[0]  # Text width
    text_height = text_size[1]  # Text height

    # Calculate position to center the text at the bottom
    x_position = (frame_width - text_width) // 2  # Center horizontally
    y_position = frame_height - text_height - 10  # Position 10px from the bottom

    # Add the disclaimer text to the frame
    cv2.putText(result_frame, disclaimer_text, (x_position, y_position), font, 1, (0, 0, 255), 2, cv2.LINE_AA)

    # Write the frame to output video file
    out.write(result_frame)

    # If you cannot view the frame, try saving to disk and check the saved output
    # cv2.imshow('Frame', result_frame)

    # Break the loop if the user presses the 'q' key (to stop displaying)
    # if cv2.waitKey(1) & 0xFF == ord('q'):
    #    break

    frame_count += 1

    # Optional: Add a break condition to stop after a certain number of frames for testing purposes
    if frame_count >= 100:  # Stop after processing 100 frames (adjust as needed)
        break

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

# Check the output path to see if the video is saved
print("Video saved at:", output_video_path)
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time

# Path to your trained model
model_path = '/kaggle/input/license/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Path to the input video file
video_path = '/kaggle/input/licenseplate/other/default/1/license.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Define the codec and create VideoWriter object to save output video
output_video_path = '/kaggle/working/license.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    resized_frame = cv2.resize(frame, (640, 640))

    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = frame.copy()

    # Blur the detected regions
    for box in results[0].boxes:
        label = box.cls[0].item()
        if label == 0:  # Assuming label '0' corresponds to the detected object (like license plate)
            x1, y1, x2, y2 = map(int, box.xyxy[0])  # Get bounding box coordinates

            # Ensure the bounding box coordinates are within valid image bounds
            if x1 < 0 or y1 < 0 or x2 >= frame_width or y2 >= frame_height:
                print(f"Invalid bounding box: {x1, y1, x2, y2}. Skipping blur.")
                continue

            roi = result_frame[y1:y2, x1:x2]  # Extract region of interest (ROI)

            # Check if the ROI is not empty before applying the blur
            if roi.size > 0:
                blurred_roi = cv2.GaussianBlur(roi, (25, 25), 0)  # Apply Gaussian blur
                result_frame[y1:y2, x1:x2] = blurred_roi  # Replace original region with blurred region
            else:
                print(f"Empty ROI for bounding box: {x1, y1, x2, y2}. Skipping blur.")

    # Write the frame to the output video
    out.write(result_frame)

    frame_count += 1

cap.release()
out.release()

print(f"Output video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time
import matplotlib.pyplot as plt

model_path = '/kaggle/input/license/other/default/1/weights/best.pt'
model = YOLO(model_path)

video_path = '/kaggle/input/licenseplate/other/default/1/license.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    resized_frame = cv2.resize(frame, (640, 640))

    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = frame.copy()

    # Loop through each detection and apply blur + bounding box
    for box in results[0].boxes:
        label = box.cls[0].item()
        if label == 0:  # Assuming label '0' corresponds to the detected object (e.g., license plate)
            x1, y1, x2, y2 = map(int, box.xyxy[0])  # Get bounding box coordinates

            # Ensure the bounding box coordinates are within valid image bounds
            if x1 < 0 or y1 < 0 or x2 >= frame_width or y2 >= frame_height:
                print(f"Invalid bounding box: {x1, y1, x2, y2}. Skipping.")
                continue

            # Extract the region of interest (ROI) and apply Gaussian blur
            roi = result_frame[y1:y2, x1:x2]  # Extract region of interest
            blurred_roi = cv2.GaussianBlur(roi, (25, 25), 0)  # Apply Gaussian blur
            result_frame[y1:y2, x1:x2] = blurred_roi  # Replace the original region with blurred region

            # Draw a bounding box around the detected object
            cv2.rectangle(result_frame, (x1, y1), (x2, y2), (0, 255, 0), 3)  # Green bounding box

    # Display the frame with bounding boxes and blurred regions using Matplotlib
    plt.figure(figsize=(10, 10))
    plt.imshow(cv2.cvtColor(result_frame, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for display
    plt.axis('off')  # Turn off axis
    plt.show()

    frame_count += 1

cap.release()
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time

# Path to your trained model
model_path = '/kaggle/input/blood_model/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Path to the input video file
video_path = '/kaggle/input/blood-vi/other/default/1/blood.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Define the codec and create VideoWriter object to save output video
output_video_path = '/kaggle/working/bloody.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    resized_frame = cv2.resize(frame, (640, 640))

    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = frame.copy()

    # Apply very strong blur to the detected regions
    for box in results[0].boxes:
        label = box.cls[0].item()
        if label == 0:  # Assuming label '0' corresponds to the detected object (e.g., license plate)
            x1, y1, x2, y2 = map(int, box.xyxy[0])  # Get bounding box coordinates

            # Ensure the bounding box coordinates are within valid image bounds
            if x1 < 0 or y1 < 0 or x2 >= frame_width or y2 >= frame_height:
                print(f"Invalid bounding box: {x1, y1, x2, y2}. Skipping blur.")
                continue

            roi = result_frame[y1:y2, x1:x2]  # Extract region of interest (ROI)

            # Check if the ROI is not empty before applying the blur
            if roi.size > 0:
                # Apply a very strong Gaussian blur with a large kernel size
                kernel_size = (151, 151)  # Increase kernel size for stronger blur
                sigma = 0  # Let OpenCV compute sigma based on kernel size

                # Apply multiple blurring passes for an even stronger effect
                for _ in range(3):  # Apply blur 3 times
                    roi = cv2.GaussianBlur(roi, kernel_size, sigma)

                result_frame[y1:y2, x1:x2] = roi  # Replace original region with blurred region
            else:
                print(f"Empty ROI for bounding box: {x1, y1, x2, y2}. Skipping blur.")

    # Write the frame to the output video
    out.write(result_frame)

    frame_count += 1

cap.release()
out.release()

print(f"Output video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time

model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

output_video_path = '/kaggle/working/disccc.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    resized_frame = cv2.resize(frame, (640, 640))

    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Check if any smoke-related objects are detected
    detected_smoke = False
    for result in results:
        if "smoke" in result.names:  # assuming "smoke" is a class name in your model
            detected_smoke = True
            break

    if detected_smoke:
        disclaimer_text = "Smoking is injurious to health"
        font = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 3  # Increase font scale for a bigger text size
        text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
        text_width = text_size[0]
        text_height = text_size[1]

        text_x = (frame_width - text_width) // 2
        text_y = frame_height - 20

        cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, font_scale, (0, 0, 255), 3, cv2.LINE_AA)

    out.write(cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR))

    frame_count += 1

cap.release()
out.release()

print(f"Output video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time

model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

output_video_path = '/kaggle/working/disc2.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    # Resize frame to 640x640 as required by the YOLO model
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform object detection
    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    detected_smoke = False  # Flag to check if smoke is detected

    # Debug: Check if the model is returning any detections
    for result in results:
        print(f"Detected Classes: {result.names}")  # Debug: Print class names
        
        if len(result.boxes.cls) == 0:
            print("No objects detected in this frame.")  # Debug: If no objects detected
        else:
            for idx, class_id in enumerate(result.boxes.cls):
                class_name = result.names[int(class_id)]
                print(f"Class ID: {class_id} -> Class Name: {class_name}")  # Debug: Print each detected class
                if class_name.lower() == "smoke":  # Ensure the class name is case-insensitive
                    detected_smoke = True
                    break
        if detected_smoke:
            break

    # If smoke is detected, add the disclaimer text
    if detected_smoke:
        disclaimer_text = "Smoking is injurious to health"
        font = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 3  # Increase font scale for a bigger text size
        text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
        text_width = text_size[0]
        text_height = text_size[1]

        text_x = (frame_width - text_width) // 2
        text_y = frame_height - 20

        # Add the disclaimer text only when smoke is detected
        cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, font_scale, (0, 0, 255), 3, cv2.LINE_AA)
        print("Smoke detected! Disclaimer added.")  # Debug: Confirm disclaimer is added

    out.write(cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR))  # Write frame to output

    frame_count += 1

cap.release()
out.release()

print(f"Output video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np

# Load the YOLO model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Input video path
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Output video path
output_video_path = '/kaggle/working/diii.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    # Resize frame to 640x640 as required by the YOLO model
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform object detection
    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = frame.copy()  # Use original frame for output

    detected_smoke = False  # Flag to check if smoke is detected

    # Check the results and process detected objects
    for result in results:
        print(f"Detected Classes: {result.names}")  # Debug: print the detected classes
        
        if len(result.boxes.cls) == 0:
            print("No objects detected in this frame.")  # No objects detected
        else:
            # Iterate over each detected object
            for idx, class_id in enumerate(result.boxes.cls):
                class_name = result.names[int(class_id)]
                print(f"Class ID: {class_id} -> Class Name: {class_name}")  # Print detected class names
                
                if class_name.lower() == "smoke":  # Case-insensitive comparison for smoke class
                    detected_smoke = True
                    print("Smoke detected!")  # Debug message when smoke is detected

                    # Get the bounding box of the detected object
                    box = result.boxes.xyxy[idx].cpu().numpy()  # Get box coordinates [x1, y1, x2, y2]
                    x1, y1, x2, y2 = box
                    x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)

                    # Draw the bounding box on the frame
                    cv2.rectangle(result_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

                    # Add the disclaimer text around the detected object
                    disclaimer_text = "Smoking is injurious to health"
                    font = cv2.FONT_HERSHEY_SIMPLEX
                    font_scale = 1.5  # Adjust for appropriate size
                    color = (0, 0, 255)  # Red color for text

                    # Calculate text size and position
                    text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
                    text_width, text_height = text_size
                    text_x = x1 + (x2 - x1 - text_width) // 2  # Center text horizontally on the bounding box
                    text_y = y1 - 10  # Position text just above the bounding box

                    # Add the text to the frame
                    if text_y - text_height < 0:
                        text_y = y2 + text_height + 10  # If text is going above, place it below the bounding box

                    cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, font_scale, color, 2, cv2.LINE_AA)
                    print("Disclaimer added at:", (text_x, text_y))  # Debug: confirm disclaimer added
                    break  # Break after adding disclaimer to the first detected object

        if detected_smoke:
            break  # Break out of the loop once "smoke" is detected in this frame

    # Write the processed frame to the output video
    out.write(result_frame)

    frame_count += 1

cap.release()
out.release()

print(f"Output video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np

# Load the YOLO model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Input video path
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Output video path
output_video_path = '/kaggle/working/dp.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    # Resize frame to 640x640 as required by the YOLO model
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform object detection
    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = frame.copy()  # Use original frame for output

    detected_smoke = False  # Flag to check if smoke is detected

    # Check the results and process detected objects
    for result in results:
        print(f"Detected Classes: {result.names}")  # Debug: print the detected classes
        
        if len(result.boxes.cls) == 0:
            print("No objects detected in this frame.")  # No objects detected
        else:
            # Iterate over each detected object
            for idx, class_id in enumerate(result.boxes.cls):
                class_name = result.names[int(class_id)]
                print(f"Class ID: {class_id} -> Class Name: {class_name}")  # Print detected class names
                
                if class_name.lower() == "smoke":  # Case-insensitive comparison for smoke class
                    detected_smoke = True
                    print("Smoke detected!")  # Debug message when smoke is detected

                    # Get the bounding box of the detected object
                    box = result.boxes.xyxy[idx].cpu().numpy()  # Get box coordinates [x1, y1, x2, y2]
                    x1, y1, x2, y2 = box
                    x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)

                    # Draw the bounding box on the frame
                    cv2.rectangle(result_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

                    # Add the disclaimer text around the detected object
                    disclaimer_text = "Smoking is injurious to health"
                    font = cv2.FONT_HERSHEY_SIMPLEX
                    font_scale = 1.5  # Adjust for appropriate size
                    color = (0, 0, 255)  # Red color for text

                    # Calculate text size and position
                    text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
                    text_width, text_height = text_size
                    text_x = x1 + (x2 - x1 - text_width) // 2  # Center text horizontally on the bounding box
                    text_y = y1 - 10  # Position text just above the bounding box

                    # Add the text to the frame
                    if text_y - text_height < 0:
                        text_y = y2 + text_height + 10  # If text is going above, place it below the bounding box

                    cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, font_scale, color, 2, cv2.LINE_AA)
                    print(f"Disclaimer added at: ({text_x}, {text_y})")  # Debug: confirm disclaimer added
                    break  # Break after adding disclaimer to the first detected object

        if detected_smoke:
            break  # Break out of the loop once "smoke" is detected in this frame

    # Check if the disclaimer is added to this frame
    if detected_smoke:
        print("Disclaimer added to the frame.")  # Confirm disclaimer was added

    # Write the processed frame to the output video
    out.write(result_frame)

    frame_count += 1

cap.release()
out.release()

print(f"Output video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np

# Load the YOLO model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Input video path
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Output video path
output_video_path = '/kaggle/working/omg.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

desired_fps = 10
frame_interval = int(fps // desired_fps)

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    # Resize frame to 640x640 as required by the YOLO model
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform object detection
    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = frame.copy()  # Use original frame for output

    detected_smoke = False  # Flag to check if smoke is detected

    # Check the results and process detected objects
    for result in results:
        print(f"Detected Classes: {result.names}")  # Debug: print the detected classes
        
        if len(result.boxes.cls) == 0:
            print("No objects detected in this frame.")  # No objects detected
        else:
            # Iterate over each detected object
            for idx, class_id in enumerate(result.boxes.cls):
                class_name = result.names[int(class_id)]
                print(f"Class ID: {class_id} -> Class Name: {class_name}")  # Print detected class names
                
                if class_name.lower() == "smoke":  # Case-insensitive comparison for smoke class
                    detected_smoke = True
                    print("Smoke detected!")  # Debug message when smoke is detected

                    # Get the bounding box of the detected object
                    box = result.boxes.xyxy[idx].cpu().numpy()  # Get box coordinates [x1, y1, x2, y2]
                    x1, y1, x2, y2 = box
                    x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)

                    print(f"Bounding Box for smoke: ({x1}, {y1}, {x2}, {y2})")  # Debug: Print bounding box

                    # Draw the bounding box on the frame
                    cv2.rectangle(result_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

                    # Add the disclaimer text around the detected object
                    disclaimer_text = "Smoking is injurious to health"
                    font = cv2.FONT_HERSHEY_SIMPLEX
                    font_scale = 1.5  # Adjust for appropriate size
                    color = (255, 255, 255)  # White color for text (change this to (0, 0, 255) for red)

                    # Calculate text size and position
                    text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
                    text_width, text_height = text_size
                    text_x = x1 + (x2 - x1 - text_width) // 2  # Center text horizontally on the bounding box
                    
                    # Check if there’s space above the bounding box for the text
                    if y1 - text_height - 10 >= 0:
                        text_y = y1 - 10  # Place text above the bounding box
                    else:
                        text_y = y2 + text_height + 10  # Place text below the bounding box

                    print(f"Text Position: ({text_x}, {text_y})")  # Debug: Print text position

                    # Add the text to the frame
                    cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, font_scale, color, 2, cv2.LINE_AA)
                    print(f"Disclaimer added at: ({text_x}, {text_y})")  # Debug: confirm disclaimer added
                    break  # Break after adding disclaimer to the first detected object

        if detected_smoke:
            break  # Break out of the loop once "smoke" is detected in this frame

    # Check if the disclaimer is added to this frame
    if detected_smoke:
        print("Disclaimer added to the frame.")  # Confirm disclaimer was added

    # Write the processed frame to the output video
    out.write(result_frame)

    frame_count += 1

cap.release()
out.release()

print(f"Output video saved to: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
from IPython.display import Image, display

# Load the YOLO model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Input video path
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

fps = cap.get(cv2.CAP_PROP_FPS)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    resized_frame = cv2.resize(frame, (640, 640))  # Resize for YOLO model

    # Perform object detection
    results = model.predict(source=resized_frame, conf=0.3)

    result_frame = frame.copy()  # Use original frame for output

    detected_smoke = False  # Flag to check if smoke is detected

    # Check the results and process detected objects
    for result in results:
        print(f"Detected Classes: {result.names}")  # Debug: print the detected classes
        
        if len(result.boxes.cls) == 0:
            print("No objects detected in this frame.")  # No objects detected
        else:
            # Iterate over each detected object
            for idx, class_id in enumerate(result.boxes.cls):
                class_name = result.names[int(class_id)]
                print(f"Class ID: {class_id} -> Class Name: {class_name}")  # Print detected class names
                
                if class_name.lower() == "smoke":  # Case-insensitive comparison for smoke class
                    detected_smoke = True
                    print("Smoke detected!")  # Debug message when smoke is detected

                    # Get the bounding box of the detected object
                    box = result.boxes.xyxy[idx].cpu().numpy()  # Get box coordinates [x1, y1, x2, y2]
                    x1, y1, x2, y2 = box
                    x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)

                    print(f"Bounding Box for smoke: ({x1}, {y1}, {x2}, {y2})")  # Debug: Print bounding box

                    # Draw the bounding box on the frame
                    cv2.rectangle(result_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

                    # Add the disclaimer text
                    disclaimer_text = "Smoking is injurious to health"
                    font = cv2.FONT_HERSHEY_SIMPLEX
                    font_scale = 1.5  # Adjust for appropriate size
                    color = (0, 0, 255)  # Red color for text

                    text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
                    text_width, text_height = text_size
                    text_x = x1 + (x2 - x1 - text_width) // 2  # Center text horizontally on the bounding box
                    
                    # Check if there’s space above the bounding box for the text
                    if y1 - text_height - 10 >= 0:
                        text_y = y1 - 10  # Place text above the bounding box
                    else:
                        text_y = y2 + text_height + 10  # Place text below the bounding box

                    print(f"Text Position: ({text_x}, {text_y})")  # Debug: Print text position

                    # Add the text to the frame
                    cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, font_scale, color, 2, cv2.LINE_AA)
                    print(f"Disclaimer added at: ({text_x}, {text_y})")  # Debug: confirm disclaimer added
                    break  # Break after adding disclaimer to the first detected object

        if detected_smoke:
            break  # Break out of the loop once "smoke" is detected in this frame

    # Add timestamp to the frame
    timestamp = f"Frame: {frame_count}"
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(result_frame, timestamp, (10, frame_height - 10), font, 0.8, (255, 255, 255), 2, cv2.LINE_AA)

    # Display the processed frame only if smoke is detected
    if detected_smoke:
        # Convert the frame to RGB (OpenCV uses BGR by default)
        result_frame_rgb = cv2.cvtColor(result_frame, cv2.COLOR_BGR2RGB)
        
        # Display the image in the notebook
        display(Image(data=cv2.imencode('.png', result_frame_rgb)[1].tobytes()))
        print(f"Displaying frame {frame_count} with disclaimer")

    frame_count += 1

cap.release()

print("Processed frames displayed with disclaimers.")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np

# Load the YOLO model (path to the trained model)
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Input video path (path to the video you want to process)
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get the FPS, width, and height of the video to set up the output video
fps = cap.get(cv2.CAP_PROP_FPS)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Set up the output video file and codec (MP4 format)
output_video_path = '/kaggle/working/output_video_with_disclaimer.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Resize frame for YOLO model (optional resizing)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform object detection
    results = model.predict(source=resized_frame, conf=0.3)

    # Flag to check if smoke is detected in the frame
    detected_smoke = False

    # Check the results for detected classes
    for result in results:
        if len(result.boxes.cls) == 0:
            continue  # No objects detected in this frame
        
        for idx, class_id in enumerate(result.boxes.cls):
            class_name = result.names[int(class_id)]
            
            if class_name.lower() == "smoke":  # Detect smoke object
                detected_smoke = True

                # Get the bounding box for the detected object (x1, y1, x2, y2)
                box = result.boxes.xyxy[idx].cpu().numpy()  # [x1, y1, x2, y2]
                x1, y1, x2, y2 = box
                x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)

                # Draw the bounding box on the original frame
                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

                # Add the disclaimer text ("Smoking is injurious to health")
                disclaimer_text = "Smoking is injurious to health"
                font = cv2.FONT_HERSHEY_SIMPLEX
                font_scale = 1.5  # Adjust font scale
                color = (0, 0, 255)  # Red color for text

                # Get text size to calculate position
                text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
                text_width, text_height = text_size
                text_x = x1 + (x2 - x1 - text_width) // 2  # Center text horizontally on the bounding box

                # Place the text below the bounding box
                text_y = y2 + text_height + 10  # Place text below the bounding box

                # Add the text on the frame
                cv2.putText(frame, disclaimer_text, (text_x, text_y), font, font_scale, color, 2, cv2.LINE_AA)
                break  # Only add disclaimer for the first detected smoke object in the frame

    # Write the processed frame to the output video file (with or without disclaimer)
    out.write(frame)

    frame_count += 1

# Release the video capture and writer objects
cap.release()
out.release()

print(f"Processed video saved at: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import matplotlib.pyplot as plt

# Load the YOLO model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'  # Make sure this path is correct
model = YOLO(model_path)

# Input video path
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'  # Update this path as needed
cap = cv2.VideoCapture(video_path)

# Check if the video is opened successfully
if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get video properties (fps, frame width, and height)
fps = cap.get(cv2.CAP_PROP_FPS)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Initialize the output video writer
output_video_path = '/kaggle/working/outr.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

frame_count = 0

# Process each frame in the video
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Resize frame for YOLO
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform object detection (detect smoke)
    results = model.predict(source=resized_frame, conf=0.3)  # Confidence threshold is 0.3

    detected_smoke = False  # Flag to track if smoke is detected

    # Check for detected objects and add bounding box + disclaimer for "smoke"
    for result in results:
        for idx, class_id in enumerate(result.boxes.cls):
            class_name = result.names[int(class_id)]
            if class_name.lower() == "smoke":  # Check if the detected object is "smoke"
                detected_smoke = True
                box = result.boxes.xyxy[idx].cpu().numpy()
                x1, y1, x2, y2 = box
                x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)

                # Draw bounding box around detected smoke
                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

                # Add disclaimer text below the bounding box
                disclaimer_text = "Smoking is injurious to health"
                font = cv2.FONT_HERSHEY_SIMPLEX
                font_scale = 1.5
                color = (0, 0, 255)  # Red color for the text

                # Calculate position for the text
                text_size = cv2.getTextSize(disclaimer_text, font, font_scale, 2)[0]
                text_width, text_height = text_size
                text_x = x1 + (x2 - x1 - text_width) // 2  # Center the text horizontally
                text_y = y2 + text_height + 10  # Place the text below the bounding box

                # Put the disclaimer text on the frame
                cv2.putText(frame, disclaimer_text, (text_x, text_y), font, font_scale, color, 2, cv2.LINE_AA)
                break  # Only add the disclaimer once for the first detected smoke object

    # Show the processed frame inline in the notebook for debugging
    if detected_smoke:
        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)  # Convert to RGB for notebook display
        plt.figure(figsize=(10, 10))
        plt.imshow(frame_rgb)
        plt.axis('off')  # Hide axis labels
        plt.show()

    # Write the processed frame to the output video file
    out.write(frame)

    frame_count += 1

cap.release()  # Release the video capture
out.release()  # Release the video writer

print(f"Video processing complete. Check the output video at: {output_video_path}")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
from matplotlib import pyplot as plt
import time  # Import time module to get timestamps

# Path to your trained model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file (replace with your video path)
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Desired FPS (10 FPS in this case)
desired_fps = 10
frame_interval = int(fps // desired_fps)  # Calculate how many frames to skip

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Skip frames based on the desired FPS
    if frame_count % frame_interval != 0:
        frame_count += 1
        continue
    
    # Debug print to track frame count and interval
    print(f"Processing frame {frame_count}")

    # Resize the frame for YOLO model to 640x640 (for inference)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the resized frame
    results = model.predict(source=resized_frame, conf=0.3)  # Reduce confidence threshold

    # Check if predictions are made
    if len(results) > 0 and len(results[0].boxes) > 0:
        print(f"Frame {frame_count}: Found {len(results[0].boxes)} detections.")
    else:
        print(f"Frame {frame_count}: No detections found.")

    # Plot the result with bounding boxes on the original frame (not resized)
    result_frame = results[0].plot()  # This plots on the original frame (with bounding boxes)

    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Get the current timestamp in seconds based on frame count
    timestamp_seconds = frame_count / fps  # Time in seconds based on frame count and fps
    timestamp = time.strftime('%H:%M:%S', time.gmtime(timestamp_seconds))  # Convert seconds to HH:MM:SS format

    # Add the timestamp to the frame
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(result_frame, timestamp, (10, 30), font, 1, (255, 255, 255), 2, cv2.LINE_AA)

    # Display the detected bounding boxes and coordinates for each object
    for idx, box in enumerate(results[0].boxes.xyxy):
        x1, y1, x2, y2 = box.cpu().numpy().astype(int)

        # Draw the bounding box
        cv2.rectangle(result_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

        # Display the coordinates on the frame
        coord_text = f"({x1}, {y1}), ({x2}, {y2})"  # x1, y1, x2, y2 coordinates
        cv2.putText(result_frame, coord_text, (x1, y1 - 10), font, 0.5, (255, 255, 255), 2, cv2.LINE_AA)

    # Show the processed frame inline in the notebook for debugging
    frame_rgb = cv2.cvtColor(result_frame, cv2.COLOR_BGR2RGB)  # Convert to RGB for notebook display
    plt.figure(figsize=(10, 10))
    plt.imshow(frame_rgb)
    plt.axis('off')  # Hide axis labels
    plt.show()

    frame_count += 1

# Release resources
cap.release()
cv2.destroyAllWindows()

```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time
import matplotlib.pyplot as plt

# Path to your trained model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Desired FPS (10 FPS in this case)
desired_fps = 10
frame_interval = int(fps // desired_fps)  # Calculate how many frames to skip

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Skip frames based on the desired FPS
    if frame_count % frame_interval != 0:
        frame_count += 1
        continue
    
    # Resize the frame for YOLO model to 640x640 (for inference)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the resized frame
    results = model.predict(source=resized_frame, conf=0.3)  # Reduce confidence threshold

    # Check if predictions are made
    detected_smoke = False
    if len(results) > 0 and len(results[0].boxes) > 0:
        # Loop through detected boxes and look for 'smoke' class
        for box in results[0].boxes:
            class_id = int(box.cls.cpu().numpy().item())  # Extract class ID and avoid DeprecationWarning
            class_name = results[0].names[class_id]  # Get the class name

            # Debug: Print the detected class name
            print(f"Detected class: {class_name}")

            # If "smoke" is detected
            if class_name.lower() == 'smoke':
                detected_smoke = True
                break  # Only process the first detected 'smoke' object for simplicity

    # Plot the result with bounding boxes on the original frame (not resized)
    result_frame = results[0].plot()  # This plots on the original frame (with bounding boxes)

    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Get the current timestamp in seconds based on frame count
    timestamp_seconds = frame_count / fps  # Time in seconds based on frame count and fps
    timestamp_milliseconds = int((timestamp_seconds - int(timestamp_seconds)) * 1000)  # Convert fractional seconds to milliseconds
    timestamp = time.strftime('%H:%M:%S', time.gmtime(timestamp_seconds))  # Convert seconds to HH:MM:SS format

    # Add the timestamp to the frame (in HH:MM:SS and milliseconds)
    font = cv2.FONT_HERSHEY_SIMPLEX
    timestamp_text = f"{timestamp}.{timestamp_milliseconds:03d}"  # Adding milliseconds
    cv2.putText(result_frame, timestamp_text, (10, 30), font, 1, (255, 255, 255), 2, cv2.LINE_AA)

    # Add disclaimer text if 'smoke' is detected directly on the image (bottom-center of the image)
    if detected_smoke:
        disclaimer_text = "Smoking is injurious to health"
        
        # Position the disclaimer text at the bottom center of the image
        text_size = cv2.getTextSize(disclaimer_text, font, 1, 2)[0]  # Calculate the size of the text
        text_x = (frame_width - text_size[0]) // 2  # Center horizontally
        text_y = frame_height - 20  # Position at the bottom

        # Draw the disclaimer text on the image
        cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, 1, (0, 0, 255), 2, cv2.LINE_AA)

    # Show the processed frame inline in the notebook for debugging
    frame_rgb = cv2.cvtColor(result_frame, cv2.COLOR_BGR2RGB)  # Convert to RGB for notebook display
    plt.figure(figsize=(10, 10))
    plt.imshow(frame_rgb)
    plt.axis('off')  # Hide axis labels
    plt.show()

    frame_count += 1

# Release resources
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time
import matplotlib.pyplot as plt

# Path to your trained model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Desired FPS (10 FPS in this case)
desired_fps = 10
frame_interval = int(fps // desired_fps)  # Calculate how many frames to skip

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Skip frames based on the desired FPS
    if frame_count % frame_interval != 0:
        frame_count += 1
        continue

    # Resize the frame for YOLO model to 640x640 (for inference)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the resized frame
    results = model.predict(source=resized_frame, conf=0.25)  # Adjusted confidence threshold

    detected_smoke = False  # Flag to check if smoke is detected

    # Debugging: Print detected class names
    if len(results) > 0 and len(results[0].boxes) > 0:
        for box in results[0].boxes:
            class_id = int(box.cls.cpu().numpy().item())  # Get the class ID
            class_name = results[0].names[class_id]  # Get the class name
            print(f"Detected class: {class_name}")

            # More flexible check for "smoke" detection
            if "smoke" in class_name.lower():
                detected_smoke = True
                break  

    # Draw bounding boxes on the original frame (not resized)
    result_frame = results[0].plot()  # This plots on the frame

    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Get the current timestamp in seconds based on frame count
    timestamp_seconds = frame_count / fps  
    timestamp_milliseconds = int((timestamp_seconds - int(timestamp_seconds)) * 1000)  
    timestamp = time.strftime('%H:%M:%S', time.gmtime(timestamp_seconds))  

    # Add the timestamp to the frame
    font = cv2.FONT_HERSHEY_SIMPLEX
    timestamp_text = f"{timestamp}.{timestamp_milliseconds:03d}"
    cv2.putText(result_frame, timestamp_text, (10, 30), font, 1, (255, 255, 255), 2, cv2.LINE_AA)

    # If smoke is detected, add the disclaimer
    if detected_smoke:
        disclaimer_text = "Smoking is injurious to health"
        
        # Calculate position to center the disclaimer
        text_size = cv2.getTextSize(disclaimer_text, font, 1, 2)[0]
        text_x = (frame_width - text_size[0]) // 2  
        text_y = frame_height - 50  # Moved higher for visibility

        # Add black outline for better visibility
        cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, 2, (0, 0, 0), 6, cv2.LINE_AA)  # Black outline
        cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, 2, (0, 0, 255), 3, cv2.LINE_AA)  # Red text

    # Show the processed frame inline in the notebook
    frame_rgb = cv2.cvtColor(result_frame, cv2.COLOR_BGR2RGB)  
    plt.figure(figsize=(10, 10))
    plt.imshow(frame_rgb)
    plt.axis('off')  
    plt.show()

    frame_count += 1
    if frame_count > 50:  # Limiting frames to process for testing
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import time
import matplotlib.pyplot as plt

# Path to your trained model
model_path = '/kaggle/input/blood_model/other/default/1/weights/best.pt'

# Initialize the model with the pre-trained weights
model = YOLO(model_path)

# Path to the input video file
video_path = '/kaggle/input/bloodoo/other/default/1/bloody.mp4'

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Unable to open video file.")
else:
    print("Video opened successfully.")

# Get the frame rate of the video
fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Video FPS: {fps}")

# Get the frame size (width and height)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Desired FPS (10 FPS in this case)
desired_fps = 10
frame_interval = int(fps // desired_fps)  # Calculate how many frames to skip

print(f"Processing at {desired_fps} FPS, skipping every {frame_interval} frames")

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print(f"Error: Failed to read frame {frame_count}.")
        break

    # Skip frames based on the desired FPS
    if frame_count % frame_interval != 0:
        frame_count += 1
        continue
    
    # Resize the frame for YOLO model to 640x640 (for inference)
    resized_frame = cv2.resize(frame, (640, 640))

    # Perform inference on the resized frame
    results = model.predict(source=resized_frame, conf=0.3)  # Reduce confidence threshold

    # Check if predictions are made
    detected_smoke = False
    if len(results) > 0 and len(results[0].boxes) > 0:
        # Loop through detected boxes and look for 'smoke' class
        for box in results[0].boxes:
            class_id = int(box.cls.cpu().numpy().item())  # Extract class ID and avoid DeprecationWarning
            class_name = results[0].names[class_id]  # Get the class name

            # Debug: Print the detected class name
            print(f"Detected class: {class_name}")

            # If "smoke" is detected
            if class_name.lower() == 'smoke':
                detected_smoke = True
                break  # Only process the first detected 'smoke' object for simplicity

    # Plot the result with bounding boxes on the original frame (not resized)
    result_frame = results[0].plot()  # This plots on the original frame (with bounding boxes)

    # Convert result frame from RGB to BGR (for OpenCV compatibility)
    result_frame = cv2.cvtColor(result_frame, cv2.COLOR_RGB2BGR)

    # Get the current timestamp in seconds based on frame count
    timestamp_seconds = frame_count / fps  # Time in seconds based on frame count and fps
    timestamp_milliseconds = int((timestamp_seconds - int(timestamp_seconds)) * 1000)  # Convert fractional seconds to milliseconds
    timestamp = time.strftime('%H:%M:%S', time.gmtime(timestamp_seconds))  # Convert seconds to HH:MM:SS format

    # Add the timestamp to the frame (in HH:MM:SS and milliseconds)
    font = cv2.FONT_HERSHEY_SIMPLEX
    timestamp_text = f"{timestamp}.{timestamp_milliseconds:03d}"  # Adding milliseconds
    cv2.putText(result_frame, timestamp_text, (10, 30), font, 1, (255, 255, 255), 2, cv2.LINE_AA)

    # Add disclaimer text if 'smoke' is detected directly on the image (bottom-center of the image)
    if detected_smoke:
        disclaimer_text = "Smoking is injurious to health"
        
        # Position the disclaimer text at the bottom center of the image
        text_size = cv2.getTextSize(disclaimer_text, font, 1, 2)[0]  # Calculate the size of the text
        text_x = (frame_width - text_size[0]) // 2  # Center horizontally
        text_y = frame_height - 20  # Position at the bottom

        # Draw the disclaimer text on the image
        cv2.putText(result_frame, disclaimer_text, (text_x, text_y), font, 1, (0, 0, 255), 2, cv2.LINE_AA)

    # Show the processed frame inline in the notebook for debugging
    frame_rgb = cv2.cvtColor(result_frame, cv2.COLOR_BGR2RGB)  # Convert to RGB for notebook display
    plt.figure(figsize=(10, 10))
    plt.imshow(frame_rgb)
    plt.axis('off')  # Hide axis labels
    plt.show()

    frame_count += 1

# Release resources
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
from ultralytics import YOLO

# Load YOLO smoke detection model
model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)

# Load the video file
video_path = '/kaggle/input/videoo/other/default/1/10273130-uhd_2160_4096_25fps.mp4'
cap = cv2.VideoCapture(video_path)

# Get video properties
fps = int(cap.get(cv2.CAP_PROP_FPS))
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Define the output video writer
output_video_path = '/kaggle/working/pixed_output.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # Codec for MP4
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break  # Exit if no frame is read

    # Resize frame for YOLO inference (optional, can use original size)
    resized_frame = cv2.resize(frame, (640, 640))
    
    # Run YOLO detection on the frame
    results = model.predict(source=resized_frame, conf=0.3)

    # If detections are found, apply pixelation
    if len(results[0].boxes) > 0:
        for box in results[0].boxes:
            x1, y1, x2, y2 = box.xyxy[0].int().tolist()
            
            # Ensure coordinates are within bounds
            x1, y1, x2, y2 = max(0, x1), max(0, y1), min(frame.shape[1], x2), min(frame.shape[0], y2)

            # Extract detected region
            detected_region = frame[y1:y2, x1:x2]

            # Apply pixelation (resize down and up)
            pixelated_region = cv2.resize(detected_region, (10, 10), interpolation=cv2.INTER_LINEAR)
            pixelated_region = cv2.resize(pixelated_region, (x2 - x1, y2 - y1), interpolation=cv2.INTER_NEAREST)

            # Replace original region with pixelated one
            frame[y1:y2, x1:x2] = pixelated_region

    # Write the processed frame to the output video
    out.write(frame)

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

print(f"Pixelated video saved to: {output_video_path}")
```

```python
import os
import cv2
import numpy as np
from ultralytics import YOLO
from IPython.display import FileLink


model_path = '/kaggle/input/smoke/other/default/1/weights/best.pt'
model = YOLO(model_path)


input_video_path = "/kaggle/input/blood-vi/other/default/1/blood.mp4"  # Change if needed
output_video_path = "/kaggle/working/pixed_blood.mp4"

cap = cv2.VideoCapture(input_video_path)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))


fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # Codec
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

frame_count = 0  


while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break  

    frame_count += 1  # Increment frame counter
    print(f"Processing frame {frame_count}...")

  
    results = model.predict(source=frame, conf=0.3)  
    detected_smoke = False  

   
    if len(results[0].boxes) > 0:
        for box in results[0].boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])  

            x1, y1, x2, y2 = max(0, x1), max(0, y1), min(frame_width, x2), min(frame_height, y2)

          
            detected_region = frame[y1:y2, x1:x2].copy()

            
            if detected_region.size > 0:
                small = cv2.resize(detected_region, (10, 10), interpolation=cv2.INTER_LINEAR)
                pixelated = cv2.resize(small, (x2 - x1, y2 - y1), interpolation=cv2.INTER_NEAREST)
                frame[y1:y2, x1:x2] = pixelated  # Replace the original region with pixelated version
                print(f"✅ Pixelation applied on detected area: ({x1}, {y1}, {x2}, {y2})")

            detected_smoke = True  # Mark that smoke was detected

    
    if detected_smoke:
        print("🚨 blood detected! Adding disclaimer.")  # Debugging print

        disclaimer_text = "Smoking is injurious to health"

        text_size = cv2.getTextSize(disclaimer_text, cv2.FONT_HERSHEY_SIMPLEX, 2, 3)[0]
        text_x = (frame_width - text_size[0]) // 2  # Center horizontally
        text_y = frame_height - 50  # Keep it above the bottom edge

       
        cv2.rectangle(frame, (text_x - 20, text_y - 40), (text_x + text_size[0] + 20, text_y + 10), (0, 0, 0), -1)

       
        cv2.putText(frame, disclaimer_text, (text_x, text_y), cv2.FONT_HERSHEY_SIMPLEX, 2, (0, 0, 255), 3, cv2.LINE_AA)

   
    out.write(frame)


cap.release()
out.release()
cv2.destroyAllWindows()


print(f"✅ Video saved successfully at: {output_video_path}")
FileLink(output_video_path)
```
